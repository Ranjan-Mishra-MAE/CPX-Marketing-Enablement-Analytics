# LOY-00620 UltraAVX Scene+ Offer (Sep 17-20, 2026) - Cineplex Theatre View

**Goal:** Restate Neha's test vs control results by the **Cineplex theatre** where AVX tickets were bought (the 5 targeted locations are Imagine Cinemas theatres).

**Logic follows Neha's query *See any movie - AVX*:**
* Test = members sent the EN email (sends basis) or sent and opened (opens basis); Control = Braze holdout step.
* Conversion = UltraAVX tickets (`Auditorium_Experience_ID = 2`) at Cineplex theatres, transaction and visit date Sep 17-21.
* Unit = distinct `cde_id`; % conversion = tickets / group size; incremental tickets = (test rate - control rate) x test size.

**Flow:** Parameters, then Spark pulls to pandas (sends, opens, control, transactions), then pandas analysis (member base, theatre roll-up, reconciliation, incrementality).

## Recommendations
1. **Target by engagement, not geography:** send to frequent (6+ visits) and prior-AVX Scene+ members, wherever they live.
2. **Size holdouts for the read:** at about 1.2 AVX tickets per 100 members, control groups of 950-9,221 cannot detect market-level lift. Use larger holdouts, or pool markets into one test.
3. **Measure out-of-market spill-over by design.** Most response happened at GTA theatres outside the Imagine radius.
4. **Confirm the incremental-revenue basis with Neha** (her figure implies about $28 per ticket).

```python
# ---------------------------------------------------------------
# PARAMETERS - change here, re-run all cells
# ---------------------------------------------------------------
import pandas as pd

CANVAS   = "[Week45] LOY-00620 - UltraAVX Scene Offer – Competitive Markets_Sep17"
OPEN_CUT = "2026 Sep 21 11:59PM EDT"
# Conversion window (yyyymmdd): offer Thu Sep 17 - Sun Sep 20 + Mon Sep 21 grace day (Neha's window)
TX_START, TX_END = 20260917, 20260921
# Segment ("market") = Imagine location parsed from the Braze step name
# (?i) = case-insensitive, so 'Control Group - Alliston comp' (lowercase) is captured
MARKET_RX = r"(?i)- (.+) comp$"   # 'Email EN - Lakeshore Comp' -> 'Lakeshore'

# Cineplex market definition (Step 8)
# Imagine theatre targeted per segment -> Rentrak theatre no. (lat/lon source)
IMAGINE_RENTRAK = {"Downtown Movie Lounge": 919, "Promenade": 998, "Elgin Mills": 7410,
                   "Alliston": 991895, "Lakeshore": 989965}
MARKET_RADIUS_KM = 15   # operating Cineplex theatres within this distance = segment's Cineplex market
FALLBACK_MAX_KM  = 60   # if no AVX theatre within radius, add nearest AVX theatre within this distance
```

## Step 1 - Load base tables (Spark to pandas)
**1a. Test sends** from `mkt_braze_users_messages_email_send`. Expect 1 row per user per step: Promenade 164,412 / Elgin Mills 83,470 / Alliston 46,753 / Lakeshore 27,049 / Downtown 17,103.

```python
# Test group = members sent the EN email, 1 row per Braze user; cde_id taken from message_extras JSON
# Assumption: cde_id fits INT (max 8 digits observed), same cast as Neha
df_sends = spark.sql(f"""
    SELECT DISTINCT user_id,
           CAST(get_json_object(message_extras, '$.cdeid') AS INT) AS cde_id,
           canvas_step_name
    FROM cpx_dataanalytics_gold.marketing.mkt_braze_users_messages_email_send
    WHERE canvas_name = '{CANVAS}'
      AND canvas_step_name LIKE 'Email%Comp'
""").toPandas()

# Tag each send with its Imagine segment
df_sends["market"] = df_sends["canvas_step_name"].str.extract(MARKET_RX)[0]

# Check: rows = users = cde per step (no duplicates); totals match Neha's 'No of sends'
df_sends.groupby(["canvas_step_name", "market"]).agg(
    rows=("user_id", "size"), users=("user_id", "nunique"), cde=("cde_id", "nunique"))
```

**1b. Opens** from `mkt_braze_users_messages_email_open`, up to Sep 21 11:59 PM ET. Kept as a user_id list and joined to sends in Step 2.

```python
# Users who opened any email in the canvas up to OPEN_CUT (Braze 'time' = unix seconds)
# Joined to sends on user_id later, so segment comes from the send step
df_opens = spark.sql(f"""
    SELECT DISTINCT user_id
    FROM cpx_dataanalytics_gold.marketing.mkt_braze_users_messages_email_open
    WHERE canvas_name = '{CANVAS}'
      AND time <= unix_timestamp(to_timestamp('{OPEN_CUT}', 'yyyy MMM dd hh:mma z'))
""").toPandas()

df_opens.shape   # expected ~153K opens (Neha: 45% open rate)
```

**1c. Control group** from `mkt_braze_users_canvasstep_progression` (`Control Group - <segment>` steps). `cde_id` comes from `cpx_d_customer` (Connect GUID), with `mkt_braze_scene_only_cust` as the fallback.

**Fixed:** the Alliston step is named `Control Group - Alliston comp` (lowercase). The query uses `ILIKE` and `MARKET_RX` uses `(?i)`, so it is now loaded (about 2,593 members, as in Neha's table).

```python
# Control group = holdout users who entered the 'Control Group - <segment>' canvas step (no email)
# cde_id resolved via Cineplex Connect GUID (cpx_d_customer), fallback to Scene-only lookup by braze_id
# Note: rows = users; cde < users because some users have no matched cde_id (NULL), e.g. Downtown 962 users / 950 cde
df_control = spark.sql(f"""
    SELECT e.user_id,
           COALESCE(c.cde_id, b.cde_id) AS cde_id,
           e.canvas_step_name
    FROM cpx_dataanalytics_gold.marketing.mkt_braze_users_canvasstep_progression e
    LEFT JOIN cpx_dataanalytics_gold.customer360.cpx_d_customer c ON e.external_user_id = c.cc_user_profile_guid
    LEFT JOIN data_exploration_hub.marketing_hub.mkt_braze_scene_only_cust b ON e.user_id = b.braze_id
    WHERE e.canvas_name = '{CANVAS}'
      AND e.canvas_step_name ILIKE 'Control Group%Comp'   -- ILIKE: case-insensitive, includes 'Control Group - Alliston comp'
""").toPandas()

# Tag each control member with its Imagine segment
df_control["market"] = df_control["canvas_step_name"].str.extract(MARKET_RX)[0]
df_control.groupby(["canvas_step_name", "market"]).agg(
    rows=("user_id", "size"), users=("user_id", "nunique"), cde=("cde_id", "nunique"))
```

**1d. UltraAVX ticket transactions** at Cineplex theatres, Sep 17-21, all members. Grain: `cde_id x Location_Name x visit_date`. Filtered to campaign members in Step 5.

```python
# All UltraAVX ticket transactions at Cineplex theatres in the conversion window (all members, not only campaign)
# Grain: cde_id x theatre x visit date. AVX = Auditorium_Experience_ID 2 (Is_AVXflag='Yes')
# LEFT JOIN + WHERE on T2 behaves as INNER JOIN (same as Neha's query)
df_tx = spark.sql(f"""
    SELECT T1.CDE_ID AS cde_id,
           T4.Location_Name,
           to_date(CAST(T1.VisitDateId AS STRING), 'yyyyMMdd') AS visit_date,
           SUM(T1.Quantity)    AS no_of_tickets,
           SUM(T1.Net_Revenue) AS net_revenue_tickets
    FROM cpx_dataanalytics_gold.customer360.c360_f_transaction T1
    INNER JOIN cpx_dataanalytics_gold.customer360.cpx_d_location T4 ON T4.LocationID = T1.LocationID
    LEFT JOIN cpx_dataanalytics_gold.customer360.cpx_d_session T2 ON T2.sessionskey = T1.SessionSkey
    WHERE T1.TicketFLAG = 'Yes'
      AND T1.TransactionDateid BETWEEN {TX_START} AND {TX_END}
      AND T1.VisitDateId      BETWEEN {TX_START} AND {TX_END}
      AND T2.Auditorium_Experience_ID = 2
      AND T1.CDE_ID IS NOT NULL
      AND T4.LC_Location_Type_Description = 'Cineplex Theatre'
    GROUP BY T1.CDE_ID, T4.Location_Name, to_date(CAST(T1.VisitDateId AS STRING), 'yyyyMMdd')
""").toPandas()

# Sanity check: total AVX tickets / revenue / buyers in window
df_tx[["no_of_tickets", "net_revenue_tickets"]].sum(), df_tx["cde_id"].nunique()
```

## Step 2 - Build member base
Three groups per segment: **Test - Sends**, **Test - Opens** (a subset of sends) and **Control**. In control, `size > nunique` comes from users with no matched `cde_id` (80 in total). They are excluded from the denominators. There are no duplicate `cde_id` rows.

```python
# Stack the three analysis groups per segment:
#   Test - Sends : everyone sent (Neha's 'based on sends')
#   Test - Opens : sent AND opened (Neha's 'based on opens')
#   Control      : holdout
test_sends = df_sends[["market", "cde_id"]].assign(grp="Test - Sends")
test_opens = (df_sends.merge(df_opens, on="user_id", how="inner")
              [["market", "cde_id"]].assign(grp="Test - Opens"))
control    = df_control[["market", "cde_id"]].assign(grp="Control")

df_members = pd.concat([test_sends, test_opens, control], ignore_index=True)
df_members["cde_id"] = df_members["cde_id"].astype("Int64")   # nullable int -> matches df_tx key type

# Check: size - nunique = control users with NULL cde_id (no duplicate cde rows)

df_members.groupby(["market", "grp"])["cde_id"].agg(["size", "nunique"])
```

## Step 3 - Aggregate AVX purchases to member x Cineplex theatre
Keeps the theatre dimension (Neha rolled straight up to `cde_id`). Rows with net tickets <= 0 are dropped first, as in Neha's `WHERE No_of_tickets > 0`.

```python
# Roll daily rows up to cde_id x Cineplex theatre (keeps theatre for the Cineplex-level view)
# visits = count of (theatre, visit date) rows, same as Neha's COUNT(Visit_Date); tickets > 0 drops refunds/net-zero days
df_tx["cde_id"] = df_tx["cde_id"].astype("Int64")

df_tx_theatre = (df_tx[df_tx["no_of_tickets"] > 0]
                 .groupby(["cde_id", "Location_Name"], as_index=False)
                 .agg(no_of_visits=("visit_date", "count"),
                      no_of_tickets=("no_of_tickets", "sum"),
                      net_revenue_tickets=("net_revenue_tickets", "sum")))

df_tx_theatre[["no_of_visits", "no_of_tickets", "net_revenue_tickets"]].sum()
```

## Step 4 - Group sizes (denominators)
Distinct `cde_id` per segment and group. These are the denominators for every conversion rate, including theatre-level rates.

```python
# Denominator per segment x group = distinct cde_id (Neha's 'No of sends'); used for every conversion rate
df_size = (df_members.groupby(["market", "grp"], as_index=False)["cde_id"]
           .nunique().rename(columns={"cde_id": "group_size"}))
df_size
```

## Step 5 - Conversions by Cineplex theatre
Segment x group x theatre: converters, visits, tickets, revenue and % conversion (theatre tickets / segment group size). A member buying at two theatres counts as a converter at both.

```python
# Attach each converting member's AVX purchases (by Cineplex theatre) to their segment/group
# Inner join -> only converters; NULL cde_id members drop out (no duplicate cde rows, so no double counting)
df_joined = df_members.merge(df_tx_theatre, on="cde_id", how="inner")   # row-level, keeps Neha's join behaviour

df_theatre = (df_joined.groupby(["market", "grp", "Location_Name"], as_index=False)
              .agg(converted=("cde_id", "nunique"),
                   no_of_visits=("no_of_visits", "sum"),
                   no_of_tickets=("no_of_tickets", "sum"),
                   net_revenue_tickets=("net_revenue_tickets", "sum"))
              .merge(df_size, on=["market", "grp"]))

# Theatre-level conversion = tickets at that theatre / total segment group size
df_theatre["pct_conversion"] = df_theatre["no_of_tickets"] / df_theatre["group_size"]   # Neha: tickets ÷ sends
df_theatre.sort_values(["market", "grp", "no_of_tickets"], ascending=[True, True, False])
```

## Step 6 - Reconcile to Neha's totals
Summing theatres back to segment level should reproduce Neha's tables. Current result:
* Matches tickets and revenue for all segments except the gap below. Alliston control is now loaded: 2,593 members / 14 tickets, matching Neha.
* Promenade control has 9,221 members, matching Neha.
* One small gap: Downtown Test - Sends has 249 tickets / $4,334 (Neha: 251 / $4,370). This is likely a data refresh since Neha's run.

```python
# Reconciliation: summing theatres back to segment level should reproduce Neha's tables
# how='right' keeps groups with zero conversions; a NaN ticket row would mean the group had no conversions
df_recon = (df_theatre.groupby(["market", "grp"], as_index=False)
            [["no_of_tickets", "net_revenue_tickets"]].sum()
            .merge(df_size, on=["market", "grp"], how="right")
            .sort_values(["market", "grp"]))
df_recon
```

## Step 7 - Incremental tickets by Cineplex theatre (sends basis)
`incremental = (test_rate - ctrl_rate) x test_n` per theatre, where rates use segment-level denominators, so theatre values add up to the segment total.

**Caveats:**
* Control tickets per theatre are very small (often 0-15), so theatre-level lift is directional. Read it at market level (theatres grouped near each Imagine location) rather than per theatre.

```python
# Incrementality by Cineplex theatre (sends basis):
#   rate = theatre tickets / segment group size; incremental = (test_rate - ctrl_rate) x test_n
# Caveat: control counts per theatre are tiny (often 0-15 tickets) -> theatre-level lift is directional only
# All 5 control groups are loaded (Alliston via ILIKE / (?i)), so ctrl_n is populated for every segment
sz = df_size.pivot(index="market", columns="grp", values="group_size")[["Test - Sends", "Control"]] \
            .rename(columns={"Test - Sends": "test_n", "Control": "ctrl_n"})

tix = (df_theatre[df_theatre["grp"].isin(["Test - Sends", "Control"])]
       .pivot_table(index=["market", "Location_Name"], columns="grp",
                    values="no_of_tickets", aggfunc="sum", fill_value=0)
       .rename(columns={"Test - Sends": "test_tix", "Control": "ctrl_tix"})
       .reset_index()
       .merge(sz, on="market"))

# Cast to float: pandas returns Decimal from Spark decimals
tix["test_rate"] = tix["test_tix"].astype(float) / tix["test_n"].astype(float)
tix["ctrl_rate"] = tix["ctrl_tix"].astype(float) / tix["ctrl_n"].astype(float)
tix["incremental_tickets"] = (tix["test_rate"] - tix["ctrl_rate"]) * tix["test_n"]

df_incr_theatre = tix.sort_values(["market", "incremental_tickets"], ascending=[True, False])
df_incr_theatre
```

## Step 8 - Define the Cineplex market for each Imagine segment
* **Imagine location:** Rentrak `theaters` (latest file per theatre).
* **Cineplex theatres:** operating theatres in `cpx_d_location`.
* **Market:** Cineplex theatres within `MARKET_RADIUS_KM` (straight-line haversine distance). If none has UltraAVX, the nearest AVX theatre within `FALLBACK_MAX_KM` is added.
* Markets can overlap (Promenade and Elgin Mills are about 10 km apart), so the same Cineplex theatre can sit in both. Each segment is evaluated on its own.

```python
import numpy as np

# Imagine theatre coordinates from Rentrak (dedupe to latest file per theatre)
imagine_ids = ",".join(str(v) for v in IMAGINE_RENTRAK.values())
df_imagine = spark.sql(f"""
    SELECT rentrak_theatre_no, theatre_name,
           CAST(latitude AS DOUBLE) AS lat, CAST(longitude AS DOUBLE) AS lon
    FROM cpx_dataanalytics_silver.rentrak.theaters
    WHERE country = 'CA' AND rentrak_theatre_no IN ({imagine_ids})
    QUALIFY ROW_NUMBER() OVER (PARTITION BY rentrak_theatre_no ORDER BY file_generated_datestamp DESC) = 1
""").toPandas()
df_imagine["market"] = df_imagine["rentrak_theatre_no"].map({v: k for k, v in IMAGINE_RENTRAK.items()})

# Operating Cineplex theatres with coordinates and AVX screen count
df_cpx_loc = spark.sql("""
    SELECT Location_Name, LocationID, City, Province_CD,
           Latitude AS lat, Longitude AS lon, Screen_Count_Total, Screen_Count_UltraAVX
    FROM cpx_dataanalytics_gold.customer360.cpx_d_location
    WHERE LC_Location_Type_Description = 'Cineplex Theatre'
      AND Status = 'Operating' AND Latitude IS NOT NULL
""").toPandas()

# Check: 5 Imagine theatres found; Location_Name unique (it is the join key to df_tx)
print(len(df_imagine), "Imagine theatres |", len(df_cpx_loc), "Cineplex theatres | unique names:",
      df_cpx_loc["Location_Name"].is_unique)
df_imagine
```

```python
def haversine_km(lat1, lon1, lat2, lon2):
    """Great-circle distance in km (vectorised)."""
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 6371.0 * 2 * np.arcsin(np.sqrt(a))

rows = []
for _, im in df_imagine.iterrows():
    d = df_cpx_loc.assign(market=im["market"],
                          km=haversine_km(im["lat"], im["lon"], df_cpx_loc["lat"], df_cpx_loc["lon"]))
    m = d[d["km"] <= MARKET_RADIUS_KM].assign(rule="within radius")
    # Fallback: guarantee at least one AVX theatre per market (e.g. small towns like Alliston)
    if not (m["Screen_Count_UltraAVX"] > 0).any():
        fb = (d[(d["Screen_Count_UltraAVX"] > 0) & (d["km"] <= FALLBACK_MAX_KM)]
              .nsmallest(1, "km").assign(rule="nearest AVX fallback"))
        m = pd.concat([m, fb])
    rows.append(m)

df_market_map = (pd.concat(rows)
                 [["market", "Location_Name", "City", "km", "Screen_Count_Total", "Screen_Count_UltraAVX", "rule"]]
                 .sort_values(["market", "km"]).reset_index(drop=True))
df_market_map["km"] = df_market_map["km"].round(1)

# Review: every segment should have >= 1 AVX theatre
print(df_market_map.groupby("market")["Screen_Count_UltraAVX"].apply(lambda s: (s > 0).sum()).rename("avx_theatres"))
df_market_map
```

## Step 9 - Incrementality by Cineplex market (in-market vs out-of-market)
Each segment's AVX purchases are split by theatre into **In-market** (inside the Step 8 market) and **Out-of-market**, plus a **Total**, which should reproduce Neha's segment numbers.
* Same formulas as Neha: % conversion = tickets / group size; lift = test / control - 1; incremental tickets = (test - control rate) x test size.
* `p_value` = two-proportion z-test on **converters** / group size. Values above 0.10 mean the test-vs-control difference is not statistically reliable.
* `incr_ticket_revenue` = incremental tickets (when > 0) x test net revenue per ticket. Neha's incremental revenue implies about $28 per ticket, so it likely uses a different basis (to confirm with her).

```python
from math import erfc, sqrt

# Tag each converting purchase as in / out of the segment's own Cineplex market
in_mkt = set(zip(df_market_map["market"], df_market_map["Location_Name"]))
df_joined["cpx_market"] = ["In-market" if (m, l) in in_mkt else "Out-of-market"
                           for m, l in zip(df_joined["market"], df_joined["Location_Name"])]

# Bucket-level totals; Total converters re-counted as distinct cde (a member can buy in both buckets)
bucket = (df_joined.groupby(["market", "grp", "cpx_market"], as_index=False)
          .agg(converters=("cde_id", "nunique"), tickets=("no_of_tickets", "sum"),
               revenue=("net_revenue_tickets", "sum")))
total = (df_joined.groupby(["market", "grp"], as_index=False)
         .agg(converters=("cde_id", "nunique"), tickets=("no_of_tickets", "sum"),
              revenue=("net_revenue_tickets", "sum"))
         .assign(cpx_market="Total"))

# Skeleton = every segment x group x bucket, so buckets with zero conversions are kept as 0
skeleton = df_size.merge(pd.DataFrame({"cpx_market": ["In-market", "Out-of-market", "Total"]}), how="cross")
df_mkt_base = (skeleton.merge(pd.concat([bucket, total]), on=["market", "grp", "cpx_market"], how="left")
               .fillna({"converters": 0, "tickets": 0, "revenue": 0}))
df_mkt_base[["tickets", "revenue"]] = df_mkt_base[["tickets", "revenue"]].astype(float)   # Decimal -> float
df_mkt_base
```

```python
def p_two_prop(x1, n1, x2, n2):
    """Two-sided p-value, pooled two-proportion z-test."""
    p = (x1 + x2) / (n1 + n2)
    se = sqrt(p * (1 - p) * (1 / n1 + 1 / n2)) if 0 < p < 1 else 0
    return erfc(abs(x1 / n1 - x2 / n2) / se / sqrt(2)) if se > 0 else np.nan

def lift_table(test_grp):
    """Test vs Control by segment x bucket. test_grp = 'Test - Sends' or 'Test - Opens'."""
    cols = ["group_size", "converters", "tickets", "revenue"]
    t = df_mkt_base[df_mkt_base["grp"] == test_grp].set_index(["market", "cpx_market"])[cols]
    c = df_mkt_base[df_mkt_base["grp"] == "Control"].set_index(["market", "cpx_market"])[cols]
    o = t.add_suffix("_test").join(c.add_suffix("_ctrl"))
    o["conv_test"] = o["tickets_test"] / o["group_size_test"]
    o["conv_ctrl"] = o["tickets_ctrl"] / o["group_size_ctrl"]
    o["lift"] = o["conv_test"] / o["conv_ctrl"] - 1
    o["incremental_tickets"] = (o["conv_test"] - o["conv_ctrl"]) * o["group_size_test"]
    o["incrementality_pct"] = o["incremental_tickets"] / o["tickets_test"]
    o["incr_ticket_revenue"] = o["incremental_tickets"].clip(lower=0) * o["revenue_test"] / o["tickets_test"]
    o["p_value"] = [p_two_prop(a, b, x, y) for a, b, x, y in
                    zip(o["converters_test"], o["group_size_test"], o["converters_ctrl"], o["group_size_ctrl"])]
    return o.round({"conv_test": 4, "conv_ctrl": 4, "lift": 3, "incremental_tickets": 0,
                    "incrementality_pct": 3, "incr_ticket_revenue": 0, "p_value": 3})

df_mkt_sends = lift_table("Test - Sends")
df_mkt_opens = lift_table("Test - Opens")

# Check: Total rows should match Neha's 'based on sends' (e.g. Promenade +543, Lakeshore -199 incremental)
display(df_mkt_sends.reset_index())
```

```python
def ticket_lift(d):
    t = d.loc[d["grp"] == "Test - Sends", "no_of_tickets"]
    c = d.loc[d["grp"] == "Control", "no_of_tickets"]
    diff = t.mean() - c.mean()
    se = sqrt(t.var() / len(t) + c.var() / len(c))
    return pd.Series({"incr_tickets": diff * len(t),
                      "ci_low":  (diff - 1.96 * se) * len(t),
                      "ci_high": (diff + 1.96 * se) * len(t),
                      "lift": t.mean() / c.mean() - 1,
                      "p_value": erfc(abs(diff / se) / sqrt(2))})

df_ci = pd.concat([b.groupby("market").apply(ticket_lift),
                   b.assign(market="All markets").groupby("market").apply(ticket_lift)])
df_ci.round({"incr_tickets": 0, "ci_low": 0, "ci_high": 0, "lift": 3, "p_value": 3})
```

# Q2 - AVX business performance in campaign markets vs national
**Question:** what was the overall impact on AVX performance (attendance, capacity utilization, etc.) in these markets during the campaign, and how did it compare to national AVX?

**Approach**
* **Source:** `c360_f_session` (1 row per session) joined to `cpx_d_session` (AVX flag) and `cpx_d_location`. All attendance here is total box office, not just Scene+ members.
* **Scopes:**
  * Each segment's Cineplex market = the AVX theatres from Step 8.
  * **All campaign markets** = the union of those theatres (Promenade and Elgin Mills overlap, so it is deduplicated).
  * **National** = every other Cineplex theatre with AVX sessions.
* **Periods:**
  * **Offer** = Thu-Sun Sep 17-20.
  * **Prior week** = Thu-Sun Sep 10-13.
  * **4-week avg** = the same Thu-Sun across the 4 prior weeks (includes the Labour Day weekend, Sep 3-6).
* **Metrics:**
  * AVX shows, AVX attendance (`seats_sold`) and AVX gross.
  * **Capacity utilization** = seats sold / total seats (`seats_total` = sold + available + held + house).
  * **ATP** = gross / attendance.
  * **AVX share** = AVX attendance / all attendance at the same theatres.
* **Read-out:** change vs prior (%, or percentage points for ratios), then **gap vs national** = market change minus national change. The gap strips out film-slate effects common to all theatres.

```python
from datetime import date, timedelta

# ---------------- Q2 PARAMETERS ----------------
OFFER_START, OFFER_END = date(2026, 9, 17), date(2026, 9, 20)   # Thu-Sun offer period
PRIOR_SHIFT_DAYS = 7          # week prior = same weekdays 7 days earlier
BASELINE_WEEKS   = 4          # extra baseline: avg of same weekdays over N prior weeks
AVX_EXPERIENCE_ID = 2         # cpx_d_session.Auditorium_Experience_ID for UltraAVX (same as Step 1d)

# Derived day sets
offer_days    = pd.date_range(OFFER_START, OFFER_END)
prior_days    = offer_days - pd.Timedelta(days=PRIOR_SHIFT_DAYS)
baseline_days = pd.DatetimeIndex(np.concatenate(
    [offer_days - pd.Timedelta(days=7 * k) for k in range(1, BASELINE_WEEKS + 1)]))
PERIODS = [("Offer", offer_days, 1), ("Prior week", prior_days, 1),
           (f"{BASELINE_WEEKS}-wk avg", baseline_days, BASELINE_WEEKS)]   # (label, days, divisor)

# Load window covers the earliest baseline day through the offer end
LOAD_START = baseline_days.min().strftime("%Y%m%d")
LOAD_END   = OFFER_END.strftime("%Y%m%d")
print("Load", LOAD_START, "-", LOAD_END, "| prior:", prior_days.min().date(), "-", prior_days.max().date())
```

```python
# Theatre x show date x AVX flag; light aggregation in Spark, analysis in pandas
# Decimals cast to DOUBLE to avoid slow Decimal conversion in toPandas()
df_sess = spark.sql(f"""
    SELECT l.Location_Name,
           to_date(f.showdateid, 'yyyyMMdd')                 AS show_date,
           s.Auditorium_Experience_ID = {AVX_EXPERIENCE_ID}  AS is_avx,
           COUNT(*)                                          AS shows,        -- 1 row per session
           SUM(f.seats_sold)                                 AS seats_sold,
           SUM(f.seats_total)                                AS seats_total,
           CAST(SUM(f.gross_revenue) AS DOUBLE)              AS gross_revenue,
           CAST(SUM(f.net_revenue)   AS DOUBLE)              AS net_revenue
    FROM cpx_dataanalytics_gold.customer360.c360_f_session f
    JOIN cpx_dataanalytics_gold.customer360.cpx_d_location l ON l.locationskey = f.locationskey
    JOIN cpx_dataanalytics_gold.customer360.cpx_d_session  s ON s.sessionskey  = f.sessionskey
    WHERE f.showdateid BETWEEN '{LOAD_START}' AND '{LOAD_END}'
      AND l.LC_Location_Type_Description = 'Cineplex Theatre'
    GROUP BY ALL
""").toPandas()
df_sess["show_date"] = pd.to_datetime(df_sess["show_date"])
df_sess["is_avx"] = df_sess["is_avx"].fillna(False).astype(bool)   # sessions without experience id -> non-AVX

# Check: every day loaded, AVX utilization in a plausible range (~10-30%)
chk = df_sess[df_sess["is_avx"]].groupby("show_date")[["shows", "seats_sold", "seats_total"]].sum()
chk["util"] = (chk["seats_sold"] / chk["seats_total"]).round(3)
print(len(chk), "days loaded")
chk.tail(8)
```

```python
# AVX theatres = Cineplex theatres with >= 1 AVX session in the load window
avx_theatres = set(df_sess.loc[df_sess["is_avx"] & (df_sess["shows"] > 0), "Location_Name"])

# Segment market scopes = Step 8 market theatres that actually ran AVX
mkt_avx = df_market_map[df_market_map["Location_Name"].isin(avx_theatres)]
SCOPES = {m: set(g["Location_Name"]) for m, g in mkt_avx.groupby("market")}

campaign_theatres = set().union(*SCOPES.values())   # deduped (Promenade / Elgin Mills overlap)
NATIONAL = "National (excl. campaign markets)"
SCOPES["All campaign markets"] = campaign_theatres
SCOPES[NATIONAL] = avx_theatres - campaign_theatres

# Review scope membership
pd.DataFrame([(k, len(v), ", ".join(sorted(v)) if len(v) < 10 else "...") for k, v in SCOPES.items()],
             columns=["scope", "avx_theatres", "theatres"])
```

```python
def period_metrics(theatres, days, divisor=1):
    """AVX KPIs for a theatre set over a set of days. Additive metrics are divided by `divisor`
    (number of weeks) so a multi-week baseline is comparable to a single Thu-Sun."""
    d = df_sess[df_sess["Location_Name"].isin(theatres) & df_sess["show_date"].isin(days)]
    a = d[d["is_avx"]]
    r = {"avx_shows":          a["shows"].sum(),
         "avx_attendance":     a["seats_sold"].sum(),
         "avx_capacity":       a["seats_total"].sum(),
         "avx_gross":          a["gross_revenue"].sum(),
         "theatre_attendance": d["seats_sold"].sum()}          # all formats, same theatres
    r = {k: v / divisor for k, v in r.items()}
    r["capacity_util"] = r["avx_attendance"] / r["avx_capacity"]
    r["atp"]           = r["avx_gross"] / r["avx_attendance"]
    r["avx_share"]     = r["avx_attendance"] / r["theatre_attendance"]
    r["attendance_per_show"] = r["avx_attendance"] / r["avx_shows"]
    return r

df_q2_long = pd.DataFrame([{"scope": s, "period": p, "avx_theatres": len(th), **period_metrics(th, days, div)}
                           for s, th in SCOPES.items() for p, days, div in PERIODS])

# Levels table (one row per scope x period)
show = df_q2_long.copy()
show[["capacity_util", "avx_share"]] = (show[["capacity_util", "avx_share"]] * 100).round(1)   # -> %
display(show.round({"avx_shows": 0, "avx_attendance": 0, "avx_capacity": 0, "avx_gross": 0,
                    "theatre_attendance": 0, "atp": 2, "attendance_per_show": 1}))
```

```python
METRICS = ["avx_shows", "avx_attendance", "attendance_per_show", "capacity_util", "atp", "avx_gross", "avx_share"]
RATIO   = {"capacity_util", "avx_share"}     # compared in percentage points; others in % change

w = df_q2_long.set_index(["scope", "period"])[METRICS].unstack("period")

def change_vs(base_period):
    """Offer vs base period: % change (pp for ratios), and the same change minus national's change."""
    o, b = w.xs("Offer", axis=1, level="period"), w.xs(base_period, axis=1, level="period")
    ch = pd.DataFrame({m: (o[m] - b[m]) * 100 if m in RATIO else (o[m] / b[m] - 1) * 100 for m in METRICS})
    gap = ch.sub(ch.loc[NATIONAL], axis=1)
    order = list(SCOPES)   # keep segment order, then All campaign, then National
    return ch.loc[order].round(1), gap.loc[order].round(1)

df_q2_chg_prior, df_q2_gap_prior = change_vs("Prior week")
df_q2_chg_base,  df_q2_gap_base  = change_vs(f"{BASELINE_WEEKS}-wk avg")

print("Offer vs prior week: % change (capacity_util / avx_share in pp)")
display(df_q2_chg_prior.reset_index())
print("Gap vs national (market change minus national change)")
display(df_q2_gap_prior.reset_index())
```

# Q3 - Why did Promenade outperform? Market-level factors
**Question:** Promenade delivered most of the incremental tickets. Is that just more sends, or are market factors at play (guest profile, showtime strategy, seating, AVX history)?

**Factors compared across the 5 segments:**
1. **Exposure:** sends, open rate, holdout size.
2. **Guest profile** (`cpx_d_customer` plus 12 months of ticket history before the offer):
   * AVX buyers, visit frequency, home-theatre distance (from `band`), French speakers, CineClub, age.
   * Profile is also compared test vs control, to check whether the groups were balanced.
3. **Who responded:** incrementality split by AVX history and visit frequency.
4. **Supply in the Cineplex market:** AVX screens, shows, prime-time share (shows at 6 PM or later), capacity utilization, top AVX title, showtime strategy.
5. **AVX history:** 12-month AVX utilization, AVX share of attendance, trend in the last 3 months vs the prior 9.
6. **Competition:** competitor theatres and screens near each Imagine location (Rentrak).

```python
# ---------------- Q3 / Q4 PARAMETERS ----------------
HIST_START, HIST_END = 20250917, 20260916    # 12-month guest history window before the offer (VisitDateId)
TREND_START          = "20250901"            # monthly AVX trend start (showdateid)
PRIME_TIME_HOUR      = 18                    # show starting >= 18:00 = prime time
FREQ_BANDS           = [-1, 0, 2, 5, 1e9]    # 12-month visit bands: 0 / 1-2 / 3-5 / 6+
FREQ_LABELS          = ["0", "1-2", "3-5", "6+"]
SEGMENTS             = ["Promenade", "Downtown Movie Lounge", "Elgin Mills", "Alliston", "Lakeshore"]   # display order

# Rentrak market radius per segment = Step 8 radius, widened to the fallback distance where used (Alliston)
SEG_RADIUS_KM = df_market_map.groupby("market")["km"].max().clip(lower=MARKET_RADIUS_KM).to_dict()
SEG_RADIUS_KM
```

```python
# Test (sends) + control members -> Spark temp view for the history join
spark.createDataFrame(
    df_members.loc[df_members["grp"].isin(["Test - Sends", "Control"]), ["market", "grp", "cde_id"]]
    .dropna().drop_duplicates().astype({"cde_id": "int64"})
).createOrReplaceTempView("q3_members")

# One row per member: demographics + 12-month ticket visits (all formats) and AVX visits
# Visits = distinct visit days with >= 1 ticket; AVX = Auditorium_Experience_ID 2
df_profile = spark.sql(f"""
    WITH cust AS (
        SELECT CAST(cde_id AS BIGINT) AS cde_id, age_cohort_id, gender_text, language_text, cineclub_subscriber_flag
        FROM cpx_dataanalytics_gold.customer360.cpx_d_customer
        WHERE cde_id IS NOT NULL
        QUALIFY ROW_NUMBER() OVER (PARTITION BY CAST(cde_id AS BIGINT) ORDER BY cc_user_profile_guid) = 1   -- 1 row per cde
    ),
    hist AS (
        SELECT CAST(t.CDE_ID AS BIGINT) AS cde_id,
               COUNT(DISTINCT t.VisitDateId) AS visits_12m,
               COUNT(DISTINCT CASE WHEN s.Auditorium_Experience_ID = {AVX_EXPERIENCE_ID} THEN t.VisitDateId END) AS avx_visits_12m
        FROM cpx_dataanalytics_gold.customer360.c360_f_transaction t
        JOIN (SELECT DISTINCT cde_id FROM q3_members) m ON m.cde_id = t.CDE_ID
        LEFT JOIN cpx_dataanalytics_gold.customer360.cpx_d_session s ON s.sessionskey = t.SessionSkey
        WHERE t.TicketFLAG = 'Yes' AND t.Quantity > 0
          AND t.VisitDateId BETWEEN {HIST_START} AND {HIST_END}
        GROUP BY 1
    )
    SELECT m.market, m.grp, m.cde_id, c.age_cohort_id, c.gender_text, c.language_text, c.cineclub_subscriber_flag,
           COALESCE(h.visits_12m, 0) AS visits_12m, COALESCE(h.avx_visits_12m, 0) AS avx_visits_12m
    FROM q3_members m
    LEFT JOIN cust c ON c.cde_id = m.cde_id
    LEFT JOIN hist h ON h.cde_id = m.cde_id
""").toPandas()

# Derived flags; home-distance band reused from the earlier home-theatre analysis (`band`)
df_profile["cde_id"] = df_profile["cde_id"].astype("Int64")
df_profile["avx_user_12m"] = df_profile["avx_visits_12m"] > 0
df_profile["freq_band"] = pd.cut(df_profile["visits_12m"], FREQ_BANDS, labels=FREQ_LABELS)
df_profile = df_profile.merge(band[["market", "grp", "cde_id", "home_band"]], on=["market", "grp", "cde_id"], how="left")

# Check: 1 row per market x grp x cde (rows should equal Test - Sends + Control sizes)
print(len(df_profile), "rows | dup keys:", df_profile.duplicated(["market", "grp", "cde_id"]).sum())
df_profile.groupby(["market", "grp"]).size().unstack()
```

```python
def profile_stats(d):
    """Share-of-members profile metrics for one segment x group."""
    top_age = d["age_cohort_id"].value_counts(normalize=True)
    return pd.Series({
        "members":              len(d),
        "pct_avx_buyer_12m":    d["avx_user_12m"].mean() * 100,
        "avg_visits_12m":       d["visits_12m"].mean(),
        "pct_6plus_visits":     (d["freq_band"] == "6+").mean() * 100,
        "pct_no_visit_12m":     (d["visits_12m"] == 0).mean() * 100,
        "pct_home_within_15km": (d["home_band"] == "<=15 km").mean() * 100,
        "pct_french":           d["language_text"].fillna("").str.lower().str.startswith("fr").mean() * 100,
        "pct_cineclub":         d["cineclub_subscriber_flag"].astype(str).str.lower().isin(["y", "yes", "true", "1"]).mean() * 100,
        "top_age_cohort":       f"{top_age.index[0]} ({top_age.iloc[0]*100:.0f}%)" if len(top_age) else None})

df_q3_profile = df_profile.groupby(["market", "grp"]).apply(profile_stats)

# Exposure (from Step 4 sizes) + Q1 totals
sz = df_size.pivot(index="market", columns="grp", values="group_size")
tot = df_mkt_sends.xs("Total", level="cpx_market")
df_q3_score = pd.DataFrame({
    "sends":               sz["Test - Sends"],
    "open_rate_pct":       sz["Test - Opens"] / sz["Test - Sends"] * 100,
    "holdout":             sz["Control"],
    "lift_pct":            tot["lift"] * 100,
    "incremental_tickets": tot["incremental_tickets"],
    "p_value":             tot["p_value"],
}).join(df_q3_profile.xs("Test - Sends", level="grp").drop(columns="members")).loc[SEGMENTS]
display(df_q3_score.round(1).reset_index())

# Balance check: test vs control on key profile metrics (large gaps = groups not comparable)
bal = df_q3_profile[["pct_avx_buyer_12m", "avg_visits_12m", "pct_6plus_visits", "pct_home_within_15km"]].astype(float).unstack("grp")
bal = pd.concat({m: bal[m][["Test - Sends", "Control"]] for m in bal.columns.levels[0]}, axis=1).loc[SEGMENTS]
print("Balance check: Test - Sends vs Control")
display(bal.round(1).reset_index())
```

```python
# Member-level AVX tickets in the conversion window (0 if no purchase), sends basis + control
tix_m = (df_joined[df_joined["grp"].isin(["Test - Sends", "Control"])]
         .groupby(["market", "grp", "cde_id"], as_index=False)["no_of_tickets"].sum())
tix_m["no_of_tickets"] = tix_m["no_of_tickets"].astype(float)
pm = df_profile.merge(tix_m, on=["market", "grp", "cde_id"], how="left").fillna({"no_of_tickets": 0})

def lift_by(dim):
    """Test vs control tickets per member within each segment x dim cell."""
    g = pm.groupby(["market", dim, "grp"], observed=True).agg(n=("cde_id", "size"), tickets=("no_of_tickets", "sum")).unstack("grp")
    o = pd.DataFrame({"n_test": g[("n", "Test - Sends")], "n_ctrl": g[("n", "Control")],
                      "tix_test": g[("tickets", "Test - Sends")], "tix_ctrl": g[("tickets", "Control")]})
    o["conv_test"] = o["tix_test"] / o["n_test"]
    o["conv_ctrl"] = o["tix_ctrl"] / o["n_ctrl"]
    o["lift_pct"] = (o["conv_test"] / o["conv_ctrl"] - 1) * 100
    o["incremental_tickets"] = (o["conv_test"] - o["conv_ctrl"]) * o["n_test"]
    o["pct_of_segment_incr"] = o["incremental_tickets"] / o.groupby(level="market")["incremental_tickets"].transform("sum") * 100
    return o.round({"conv_test": 4, "conv_ctrl": 4, "lift_pct": 0, "incremental_tickets": 0, "pct_of_segment_incr": 0})

df_q3_by_avx  = lift_by("avx_user_12m")   # True = bought AVX in the 12 months before the offer
df_q3_by_freq = lift_by("freq_band")
print("By prior AVX buyer (12m)"); display(df_q3_by_avx.reset_index())
print("By 12-month visit frequency"); display(df_q3_by_freq.reset_index())
```

```python
# AVX sessions for prior week + offer: theatre x date x start hour x title (+ theatre showtime strategy)
df_avx_ses = spark.sql(f"""
    SELECT l.Location_Name, l.Showtime_Strategy_Description AS showtime_strategy,
           to_date(f.showdateid, 'yyyyMMdd') AS show_date, hour(s.Show_Datetime) AS show_hour,
           fm.title_name, COUNT(*) AS shows, SUM(f.seats_sold) AS seats_sold, SUM(f.seats_total) AS seats_total
    FROM cpx_dataanalytics_gold.customer360.c360_f_session f
    JOIN cpx_dataanalytics_gold.customer360.cpx_d_location l ON l.locationskey = f.locationskey
    JOIN cpx_dataanalytics_gold.customer360.cpx_d_session  s ON s.sessionskey  = f.sessionskey
    LEFT JOIN cpx_dataanalytics_gold.customer360.cpx_d_film fm ON fm.filmskey = f.filmskey
    WHERE f.showdateid BETWEEN '{prior_days.min():%Y%m%d}' AND '{LOAD_END}'
      AND s.Auditorium_Experience_ID = {AVX_EXPERIENCE_ID}
      AND l.LC_Location_Type_Description = 'Cineplex Theatre'
    GROUP BY ALL
""").toPandas()
df_avx_ses["show_date"] = pd.to_datetime(df_avx_ses["show_date"])

# Monthly AVX vs all-format attendance by theatre, 12 months to the offer (AVX history / trend)
df_month = spark.sql(f"""
    SELECT l.Location_Name, date_trunc('MONTH', to_date(f.showdateid, 'yyyyMMdd')) AS month,
           s.Auditorium_Experience_ID = {AVX_EXPERIENCE_ID} AS is_avx,
           SUM(f.seats_sold) AS seats_sold, SUM(f.seats_total) AS seats_total
    FROM cpx_dataanalytics_gold.customer360.c360_f_session f
    JOIN cpx_dataanalytics_gold.customer360.cpx_d_location l ON l.locationskey = f.locationskey
    JOIN cpx_dataanalytics_gold.customer360.cpx_d_session  s ON s.sessionskey  = f.sessionskey
    WHERE f.showdateid BETWEEN '{TREND_START}' AND '{(OFFER_START - timedelta(days=1)):%Y%m%d}'
      AND l.LC_Location_Type_Description = 'Cineplex Theatre'
    GROUP BY ALL
""").toPandas()
df_month["month"] = pd.to_datetime(df_month["month"])
df_month["is_avx"] = df_month["is_avx"].fillna(False).astype(bool)
print(df_avx_ses.shape, df_month["month"].min().date(), "-", df_month["month"].max().date())
```

```python
loc_info = df_cpx_loc.set_index("Location_Name")
last3 = sorted(df_month["month"].unique())[-3:]          # last 3 full/partial months before the offer

def supply_stats(theatres, seg_sends=None):
    o = df_avx_ses[df_avx_ses["Location_Name"].isin(theatres) & df_avx_ses["show_date"].isin(offer_days)]
    p = df_avx_ses[df_avx_ses["Location_Name"].isin(theatres) & df_avx_ses["show_date"].isin(prior_days)]
    titles = o.groupby("title_name")["seats_sold"].sum().sort_values(ascending=False)
    mh = df_month[df_month["Location_Name"].isin(theatres)]
    avx_m, all_m = mh[mh["is_avx"]], mh
    r = {
        "avx_theatres":         len(theatres),
        "avx_screens":          loc_info.reindex(list(theatres))["Screen_Count_UltraAVX"].sum(),
        "avx_shows_per_day":    o["shows"].sum() / len(offer_days),
        "avx_seats_per_day":    o["seats_total"].sum() / len(offer_days),
        "pct_prime_time_shows": o.loc[o["show_hour"] >= PRIME_TIME_HOUR, "shows"].sum() / o["shows"].sum() * 100,
        "util_prior_pct":       p["seats_sold"].sum() / p["seats_total"].sum() * 100,
        "util_offer_pct":       o["seats_sold"].sum() / o["seats_total"].sum() * 100,
        "top_avx_title":        f"{titles.index[0]} ({titles.iloc[0] / titles.sum() * 100:.0f}%)" if len(titles) else None,
        "showtime_strategy":    ", ".join(sorted(o["showtime_strategy"].dropna().unique())),
        "avx_util_12m_pct":     avx_m["seats_sold"].sum() / avx_m["seats_total"].sum() * 100,
        "avx_share_12m_pct":    avx_m["seats_sold"].sum() / all_m["seats_sold"].sum() * 100,
        # Trend: AVX share of attendance, last 3 months vs prior months (pp)
        "avx_share_trend_pp":  (avx_m.loc[avx_m["month"].isin(last3), "seats_sold"].sum() / all_m.loc[all_m["month"].isin(last3), "seats_sold"].sum()
                                - avx_m.loc[~avx_m["month"].isin(last3), "seats_sold"].sum() / all_m.loc[~all_m["month"].isin(last3), "seats_sold"].sum()) * 100}
    if seg_sends is not None:   # exposure intensity: sends per AVX seat offered in the market over the offer weekend
        r["sends_per_avx_seat"] = seg_sends / o["seats_total"].sum()
    return pd.Series(r)

df_q3_supply = pd.DataFrame({s: supply_stats(SCOPES[s], sz.loc[s, "Test - Sends"]) for s in SEGMENTS}).T
df_q3_supply.loc[NATIONAL] = supply_stats(SCOPES[NATIONAL])
display(df_q3_supply.reset_index().rename(columns={"index": "scope"}).round(2))
```

### Q3 findings - what drives Promenade's result
**1. Scale.** Promenade had 164K sends, about 2-10x the other segments, with a similar open rate (44% vs 44-48%). Exposure *relative to local AVX capacity* was no higher: 3.9 sends per AVX seat offered, vs 3.9 at Downtown and Elgin Mills and 7.7-9.2 at Alliston and Lakeshore.

**2. A much more engaged guest base (main market-specific factor).** These are test members' activity in the 12 months before the offer:

| | Promenade | Other 4 segments |
| --- | --- | --- |
| Bought AVX | 28.8% | 13-23% |
| Avg visits | 4.7 | 2.1-2.9 |
| 6+ visits | 26% | 9-15% |
| No visit | 18% | 25-38% |
| CineClub | 14% | 5-8% |

**3. The lift comes from frequent moviegoers.**
* Promenade members with 6+ visits generated +530 of its +543 incremental tickets (lift +37%).
* Frequent guests also responded at Downtown (+40%), Elgin Mills (+7%) and Alliston (+5%). The exception is Lakeshore (-43%, on 16 holdout tickets).
* Lighter and non-AVX guests show no consistent lift.
* Promenade has the largest frequent-guest pool (42.8K members with 6+ visits), so it produces the most incremental tickets.

**4. Not driven by supply, programming or seating.**
* All markets run *Show Strategy A* and have a similar prime-time share (43-53% of AVX shows; national 48%).
* The same title dominated AVX everywhere (Resident Evil: 73% of Promenade AVX seats, 92% nationally).
* Capacity was not a constraint. Promenade AVX utilization was 5.5% in the prior week and 12.0% on the offer weekend (national 16.9%).

**5. AVX history is average.** Promenade's 12-month AVX utilization is 13.6% (national 14.5%) and its AVX share of attendance is 16.9% (national 16.1%). Its AVX share grew less in the last 3 months (+1.3 pp vs national +2.9 pp).

**6. Competition: Cineplex is dominant around Promenade** (8 theatres / 88 screens vs Imagine 6 screens plus 5 other theatres / 24 screens). Lakeshore is the most contested: 1 Cineplex (12 screens) vs Imagine (10) plus 4 others (12).

**Balance check:** test and control profiles match closely in every segment (e.g. Lakeshore AVX buyers 18.0% vs 18.2%). Lakeshore's -44% is therefore holdout noise, not a profile imbalance.

**Implication:** target frequent and AVX-engaged Scene+ members, wherever they live, rather than everyone near the Imagine theatre. Subgroup control counts are small, so treat the split as directional.

# Q4 - Cineplex box-office market share: campaign markets vs national
**Question:** did Cineplex share in the targeted markets move vs national? (Offer week vs week prior.)

**Approach**
* **Source:** Rentrak `flashrevenue` (all exhibitors), deduped to the latest file per theatre, title and day, Canada only. The archive table ends in 2025, so it is not needed.
* **Market:** all Rentrak theatres within the segment radius of the Imagine theatre (15 km; 33 km for Alliston). Theatre roles come from the Q3 competition cell.
* **Periods:** Offer Thu-Sun Sep 17-20 vs Prior Thu-Sun Sep 10-13.
* **Metrics:**
  * **Cineplex share** = Cineplex box office / total market box office.
  * **Imagine share** = the targeted Imagine theatre's share.
  * **Gap vs national** = market share change minus national share change (pp).
* **Caveat:** Rentrak has no format split, so this is total box office, not AVX-only.

```python
# Theatre x day box office for prior week + offer; dedupe to latest file per theatre/title/day
df_rt_rev = spark.sql(f"""
    WITH fr AS (
        SELECT rentrak_theatre_no, rentrak_title_no, exhibition_date, revenue
        FROM cpx_dataanalytics_silver.rentrak.flashrevenue
        WHERE exhibition_date BETWEEN '{prior_days.min():%Y-%m-%d}' AND '{OFFER_END:%Y-%m-%d}'
        QUALIFY ROW_NUMBER() OVER (PARTITION BY rentrak_theatre_no, rentrak_title_no, exhibition_date
                                   ORDER BY file_generated_datestamp DESC) = 1
    )
    SELECT rentrak_theatre_no, to_date(exhibition_date) AS exhibition_date, SUM(revenue) AS revenue
    FROM fr GROUP BY ALL
""").toPandas()
df_rt_rev["exhibition_date"] = pd.to_datetime(df_rt_rev["exhibition_date"])

# Canada only (theatres in df_rt) and label periods; days between periods (Sep 14-16) dropped
df_rt_rev = df_rt_rev.merge(df_rt[["rentrak_theatre_no", "is_cineplex"]], on="rentrak_theatre_no", how="inner")
df_rt_rev["period"] = np.select([df_rt_rev["exhibition_date"].isin(offer_days),
                                 df_rt_rev["exhibition_date"].isin(prior_days)], ["Offer", "Prior week"], "")
df_rt_rev = df_rt_rev[df_rt_rev["period"] != ""]

# Check: national Cineplex share by period (sanity: expect a stable share between weeks)
chk = df_rt_rev.groupby(["period", "is_cineplex"])["revenue"].sum().unstack()
chk["cineplex_share_pct"] = (chk[True] / chk.sum(axis=1) * 100).round(1)
chk
```

```python
def share_rows(scope, d, imagine_ids):
    """Cineplex / Imagine-target box office and share for a set of theatres, per period."""
    out = []
    for p in ["Prior week", "Offer"]:
        x = d[d["period"] == p]
        out.append({"scope": scope, "period": p,
                    "theatres_reporting": x["rentrak_theatre_no"].nunique(),
                    "total_box_office":   x["revenue"].sum(),
                    "cineplex_box_office": x.loc[x["is_cineplex"], "revenue"].sum(),
                    "imagine_box_office":  x.loc[x["rentrak_theatre_no"].isin(imagine_ids), "revenue"].sum()})
    return out

rows = []
for seg in SEGMENTS:
    ids = set(df_rt_market.loc[df_rt_market["market"] == seg, "rentrak_theatre_no"])
    rows += share_rows(seg, df_rt_rev[df_rt_rev["rentrak_theatre_no"].isin(ids)], {IMAGINE_RENTRAK[seg]})
camp_ids = set(df_rt_market["rentrak_theatre_no"])     # union of all segment markets (deduped)
rows += share_rows("All campaign markets", df_rt_rev[df_rt_rev["rentrak_theatre_no"].isin(camp_ids)], set(IMAGINE_RENTRAK.values()))
rows += share_rows(NATIONAL, df_rt_rev[~df_rt_rev["rentrak_theatre_no"].isin(camp_ids)], set())

df_q4_long = pd.DataFrame(rows)
df_q4_long["cineplex_share_pct"] = df_q4_long["cineplex_box_office"] / df_q4_long["total_box_office"] * 100
df_q4_long["imagine_share_pct"]  = df_q4_long["imagine_box_office"]  / df_q4_long["total_box_office"] * 100

w4 = df_q4_long.set_index(["scope", "period"]).unstack("period")
df_q4 = pd.DataFrame({
    "cpx_share_prior":           w4[("cineplex_share_pct", "Prior week")],
    "cpx_share_offer":           w4[("cineplex_share_pct", "Offer")],
    "imagine_share_prior":       w4[("imagine_share_pct", "Prior week")],
    "imagine_share_offer":       w4[("imagine_share_pct", "Offer")],
    "market_box_office_chg_pct": (w4[("total_box_office", "Offer")] / w4[("total_box_office", "Prior week")] - 1) * 100,
    "cpx_box_office_chg_pct":    (w4[("cineplex_box_office", "Offer")] / w4[("cineplex_box_office", "Prior week")] - 1) * 100,
})
df_q4["cpx_share_chg_pp"]     = df_q4["cpx_share_offer"] - df_q4["cpx_share_prior"]
df_q4["imagine_share_chg_pp"] = df_q4["imagine_share_offer"] - df_q4["imagine_share_prior"]
df_q4["cpx_gap_vs_national_pp"] = df_q4["cpx_share_chg_pp"] - df_q4.loc[NATIONAL, "cpx_share_chg_pp"]
df_q4 = df_q4.loc[SEGMENTS + ["All campaign markets", NATIONAL]]
display(df_q4.round(1).reset_index())
```
