```sql
Select  DATE_FORMAT(to_date(CAST(VisitDateId AS STRING), 'yyyyMMdd'), 'MMMM') AS Visit_Month, CASE CineClub_Discount_Type_ID
    WHEN 5 THEN '9.99 Companion Tickets'
    WHEN 6 THEN '9.99 Free Ticket Redemptions'
    WHEN 7 THEN '10.99 Companion Tickets'
    WHEN 8 THEN '10.99 Free Ticket Redemptions'
  END AS Discount_Type,
  sum(Quantity)from cpx_dataanalytics_gold.customer360.c360_f_transaction where VisitDateId > 20260701 and VisitDateId < 20260831 and CineClub_Discount_Type_ID in (5,6,7,8) group by DATE_FORMAT(to_date(CAST(VisitDateId AS STRING), 'yyyyMMdd'), 'MMMM'),
  MONTH(to_date(CAST(VisitDateId AS STRING), 'yyyyMMdd')),
  CASE CineClub_Discount_Type_ID
    WHEN 5 THEN '9.99 Companion Tickets'
    WHEN 6 THEN '9.99 Free Ticket Redemptions'
    WHEN 7 THEN '10.99 Companion Tickets'
    WHEN 8 THEN '10.99 Free Ticket Redemptions'
  END order by MONTH(to_date(CAST(VisitDateId AS STRING), 'yyyyMMdd')), 2
```
