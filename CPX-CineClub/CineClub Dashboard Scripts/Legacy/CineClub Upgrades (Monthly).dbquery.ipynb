```sql
SELECT
  CAST(from_unixtime(record_occurred_at) AS DATE) AS day,
  count(distinct cc_user_profile_id) AS total_users
FROM cpx_dataanalytics_gold.marketing.mkt_subscription_events
WHERE event_type = 'subscription_changed'
  AND changed = true
  AND changed_to = 'annual1'
  AND CAST(from_unixtime(record_occurred_at) AS DATE) = date_sub(current_date(), 2)
GROUP BY day
ORDER BY day
```
