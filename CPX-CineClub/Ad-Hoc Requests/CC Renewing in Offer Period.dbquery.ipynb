```sql
SELECT
    CASE
        WHEN to_date(renewal_datetime) BETWEEN '2026-10-26' AND '2026-11-08' THEN '1. Oct 26 - Nov 8'
        WHEN to_date(renewal_datetime) BETWEEN '2026-11-27' AND '2026-11-30' THEN '2. Nov 27 - 30'
        WHEN to_date(renewal_datetime) BETWEEN '2026-12-11' AND '2026-12-31' THEN '3. Dec 11 - 31'
        ELSE 'Outside windows'
    END AS offer_period,
    COUNT(DISTINCT cc_user_profile_id) AS members
FROM cpx_dataanalytics_gold.marketing.mkt_subscriber_profile_snapshot
WHERE snapshot_date_id = '20260927'
  AND plan_id = 'annual1'
  AND renewal_status = 'Active'
GROUP BY 1
ORDER BY 1;
```
