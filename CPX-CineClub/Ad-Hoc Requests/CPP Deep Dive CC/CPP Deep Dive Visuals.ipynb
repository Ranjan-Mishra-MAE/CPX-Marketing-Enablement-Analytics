```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# ── Cineplex Brand Palette ────────────────────────────────────────
HERO      = "#004CCC"   # Primary blue — headlines, CC primary
VILLAIN   = "#002C6B"   # Deep navy — titles, dark contrast
SIDEKICK  = "#62B5E5"   # Light blue — supporting elements
SUNSHINE  = "#FFD500"   # Gold — highlights, Scene / accent
SHOWTIME  = "#000000"   # Black — body text
LIGHTS_ON = "#FFFFFF"   # White — backgrounds

# Legacy aliases used by chart functions
NAVY  = VILLAIN          # CC / primary series
GOLD  = SUNSHINE         # Scene / secondary series
GREEN = SIDEKICK         # "2025" in 3-year charts
GRAY  = "#8C8C8C"        # Neutral mid-tone

# Stacked-area palette: Candy, Fountain, Hot Foods, Merch, Popcorn
STACK_COLORS = [SUNSHINE, SIDEKICK, GRAY, HERO, VILLAIN]

# ── qbstyles-inspired layout ─────────────────────────────────────
FIG_SIZE  = (10, 4.5)
QB_TEXT   = "#4d4d4d"                     # Soft ink — axes, ticks, labels
TITLE_KW  = dict(fontsize=14, color=VILLAIN, weight="bold", loc="left")

plt.rcParams.update({
    "font.family":        "sans-serif",
    "font.size":          12,
    "figure.facecolor":   "white",
    "axes.facecolor":     "white",
    "axes.edgecolor":     QB_TEXT,
    "axes.linewidth":     0.5,
    "axes.labelcolor":    QB_TEXT,
    "axes.labelsize":     12,
    "axes.titlesize":     14,
    "text.color":         QB_TEXT,
    "xtick.color":        QB_TEXT,
    "ytick.color":        QB_TEXT,
    "xtick.labelsize":    11,
    "ytick.labelsize":    11,
    "xtick.major.size":   0,
    "ytick.major.size":   0,
    "xtick.minor.size":   0,
    "ytick.minor.size":   0,
    "axes.grid":          True,
    "axes.grid.axis":     "y",
    "grid.color":         "#ededed",
    "grid.linewidth":     0.5,
    "grid.alpha":         1.0,
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "lines.linewidth":    2.5,
    "lines.markersize":   8,
    "legend.frameon":     False,
    "legend.fontsize":    11,
})

MONTHS = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

def show(fig):
    """Apply layout with breathing room and display inline."""
    fig.tight_layout(pad=2.0)
    display(fig)
    plt.close(fig)
```

```python
def chart_cpp_gap():
    gap2024 = [0.12, 1.43, -0.49, 0.20, -0.32, -0.77, -0.27, -0.77, -0.36, -0.18, -0.97, -1.05]
    gap2025 = [-0.27, -0.16, -0.68, -0.43, -1.47, -1.46, -0.99, -0.13, -1.30, -1.09, -2.06, -1.22]
    gap2026 = [-1.24, -0.17, -1.49, -2.12, -2.25, -1.95, -2.07, -2.29]

    fig, ax = plt.subplots(figsize=FIG_SIZE)
    ax.axhline(0, color="#cccccc", linewidth=0.8)
    ax.plot(MONTHS, gap2024, marker='o', color=GOLD, label="2024", linewidth=2)
    ax.plot(MONTHS, gap2025, marker='o', color=GREEN, label="2025", linewidth=2)
    ax.plot(MONTHS[:8], gap2026, marker='o', color=NAVY, label="2026", linewidth=2.5)
    ax.set_ylabel("CC \u2212 Scene CPP ($)")
    ax.set_title("CineClub vs. Scene CPP Gap, by Year", **TITLE_KW)
    ax.legend(frameon=False, loc="lower left")
    show(fig)

chart_cpp_gap()
```

```python
def chart_attach_crossing():
    years = ["2024", "2025", "2026\n(Jan-Aug)"]
    cc_att = [66.38, 64.86, 63.74]
    scene_att = [61.44, 63.08, 64.58]

    fig, ax = plt.subplots(figsize=FIG_SIZE)
    ax.plot(years, cc_att, marker='o', markersize=8, color=NAVY, linewidth=2.5, label="CC")
    ax.plot(years, scene_att, marker='o', markersize=8, color=GOLD, linewidth=2.5, label="Scene")
    ax.set_ylim(60, 68)
    ax.set_ylabel("Avg. Concession Attach Rate (%)")
    ax.set_title("CC vs. Scene Attach Rate \u2014 Diverging Trend", **TITLE_KW)
    ax.legend(frameon=False, loc="center left")
    show(fig)

chart_attach_crossing()
```

```python
def chart_cc_visits():
    v2024 = [120961, 106509, 169213, 117372, 141675, 146451, 208628, 189698, 134003, 138143, 174104, 194930]
    v2025 = [146877, 149528, 142470, 183483, 235008, 221101, 299787, 230878, 184697, 166402, 228711, 255974]
    v2026 = [206765, 182301, 250254, 244899, 298044, 286638, 316370, 360797]

    fig, ax = plt.subplots(figsize=FIG_SIZE)
    ax.plot(MONTHS, v2024, marker='o', color=GOLD, linewidth=2, label="2024")
    ax.plot(MONTHS, v2025, marker='o', color=GREEN, linewidth=2, label="2025")
    ax.plot(MONTHS[:8], v2026, marker='o', color=NAVY, linewidth=2.5, label="2026")
    ax.set_ylabel("CC Visits")
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{int(x/1000)}K"))
    ax.set_title("CineClub Visits, by Year", **TITLE_KW)
    ax.legend(frameon=False, loc="upper left")
    show(fig)

chart_cc_visits()
```

```python
def chart_visits_vs_attach_2026():
    mo26 = MONTHS[:8]
    v26 = [206765, 182301, 250254, 244899, 298044, 286638, 316370, 360797]
    att26 = [64.8, 65.8, 62.9, 63.9, 63.9, 64.2, 62.7, 61.7]

    fig, ax1 = plt.subplots(figsize=FIG_SIZE)
    ax1.bar(mo26, v26, color="#B3D4F0", label="CC Visits")
    ax1.set_ylabel("CC Visits", color=NAVY)
    ax1.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{int(x/1000)}K"))
    ax2 = ax1.twinx()
    ax2.spines['right'].set_visible(True)
    ax2.plot(mo26, att26, marker='o', color=GOLD, linewidth=2.5, label="Attach Rate")
    ax2.set_ylabel("Attach Rate (%)", color=GOLD)
    ax2.grid(False)
    ax1.set_title("2026: Rising Visits, Falling Attach Rate", **TITLE_KW)
    h1, l1 = ax1.get_legend_handles_labels()
    h2, l2 = ax2.get_legend_handles_labels()
    ax1.legend(h1 + h2, l1 + l2, frameon=False, loc="upper left")
    show(fig)

chart_visits_vs_attach_2026()
```

```python
def chart_category_share():
    candy    = [13.4, 13.0, 13.6, 13.2, 13.0, 13.4, 12.9, 13.1, 13.1, 12.8, 12.5, 12.4]
    fountain = [36.0, 34.9, 35.1, 34.8, 34.6, 35.3, 34.4, 35.6, 35.3, 34.6, 34.3, 33.9]
    hotfoods = [3.8, 3.9, 3.7, 3.7, 3.8, 3.9, 3.6, 3.9, 3.8, 3.8, 3.8, 3.5]
    merch    = [2.1, 3.7, 2.6, 4.8, 4.6, 3.6, 7.1, 1.8, 2.1, 2.7, 4.3, 5.5]
    popcorn  = [44.7, 44.6, 45.0, 43.5, 43.9, 43.8, 41.9, 45.6, 45.6, 46.2, 45.2, 44.7]

    fig, ax = plt.subplots(figsize=FIG_SIZE)
    ax.stackplot(MONTHS, candy, fountain, hotfoods, merch, popcorn,
                 labels=["Candy", "Fountain", "Hot Foods", "Merch", "Popcorn"],
                 colors=STACK_COLORS, alpha=0.9)
    ax.set_ylabel("% of CC Concession Revenue")
    ax.set_title("CC Revenue Share by Category, by Month", **TITLE_KW)
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.15), ncol=5, frameon=False, fontsize=9)
    ax.set_ylim(0, 100)
    show(fig)

chart_category_share()
```

```python
def chart_merch_share_quarter():
    quarters = ["Q1", "Q2", "Q3", "Q4"]
    m2024 = [2.7, 1.9, 3.1, 3.4]
    m2025 = [3.2, 3.8, 4.4, 5.4]
    m2026 = [2.6, 7.3, 9.0]

    fig, ax = plt.subplots(figsize=FIG_SIZE)
    ax.plot(quarters, m2024, marker='o', color=GOLD, linewidth=2, label="2024")
    ax.plot(quarters, m2025, marker='o', color=GREEN, linewidth=2, label="2025")
    ax.plot(quarters[:3], m2026, marker='o', color=NAVY, linewidth=2.5, label="2026")
    ax.set_ylabel("Merch Share of CC Revenue (%)")
    ax.set_title("CC Merch Share, by Quarter \u2014 Accelerating", **TITLE_KW)
    ax.legend(frameon=False, loc="upper left")
    show(fig)

chart_merch_share_quarter()
```
