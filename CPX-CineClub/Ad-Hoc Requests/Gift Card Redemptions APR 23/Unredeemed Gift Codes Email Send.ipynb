```python
df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("/Volumes/data_exploration_hub/marketing_hub/offline_data/Coupons APR 23.csv")
    .drop("Notes")
)

from pyspark.sql import functions as F

df_gift = spark.table("cpx_dataanalytics_silver.cineclub.gifting")
df_subs = spark.table("cpx_dataanalytics_gold.marketing.mkt_subscriber_profile_snapshot").filter(F.col("snapshot_date_id") == '20260525')
```

```python
df_all = (
df.join(df_gift, df.Coupon_Code == df_gift.id, "inner")
.filter(F.col("is_redeemed") == "False")
.select("purchaser_id", "id","is_redeemed", "redeemer_id")
)
```

```python
df_cde = (
    df_all
    .join(df_subs, df_all.purchaser_id == df_subs.cc_user_profile_id, "inner")
    .select("cde_id", "id","is_redeemed", "redeemer_id").distinct())
```
