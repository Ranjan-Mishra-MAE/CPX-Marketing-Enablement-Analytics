```python
df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("/Volumes/data_exploration_hub/marketing_hub/offline_data/Coupons APR 23.csv")
    .drop("Notes")
)

from pyspark.sql import functions as F

df_gift = spark.table("cpx_dataanalytics_silver.cineclub.gifting")
```

```python
df_all = (
df.join(df_gift, df.Coupon_Code == df_gift.id, "inner")
.select("Coupon_Code", "Coupon_Id", "Coupon_Name", "Coupon_Set", "Coupon_Status", "purchaser_id")
)
```
