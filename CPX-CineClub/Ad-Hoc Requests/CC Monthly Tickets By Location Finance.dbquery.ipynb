```sql
Select 
t.LocationId,
L.Location_Name,
coalesce(Sum(Case When CineClub_Discount_Type_ID = 6 then Quantity End)) as Total_9_99_Tickets,
coalesce(Sum(Case When Cineclub_Discount_Type_ID = 8 then Quantity End)) as Total_10_99_Tickets
From cpx_dataanalytics_gold.customer360.c360_f_transaction as t
left join cpx_dataanalytics_gold.customer360.cpx_d_location as L
on t.LocationId = L.LocationID
Where TransactionDateid BETWEEN 20261001 AND 20261031
AND LC_Location_Type_Description Like "%Theatre%"
GROUP BY t.LocationId, L.Location_Name
ORDER BY t.LocationId

```
