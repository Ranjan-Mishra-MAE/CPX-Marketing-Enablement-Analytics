```python
T_gift_card = spark.table("cpx_dataanalytics_silver.cineclub.gifting")
```

```python
import pyspark.sql.functions as F
```

```python
import pandas as pd

df_excel = pd.read_excel("/Volumes/data_exploration_hub/marketing_hub/offline_data/Tims RUTW 2026 100 CineClub Codes.xlsx", header=None, names=["Gift_Code"], engine="openpyxl")

df_excel["Gift_Code"] = df_excel["Gift_Code"].str.lstrip()

df_excel = spark.createDataFrame(df_excel)
```

```python
df_gift_status = (
    df_excel.join(
        T_gift_card.select(
            F.col("id").alias("Gift_Code"),
            F.col("is_redeemed").alias("Redeemed"),
            F.to_date("redemption_date").alias("redemption_date")
        ),
        on="Gift_Code",
        how="inner"
    )
)
```
