```python
import pandas as pd
from pyspark.sql import functions as F
```

```python
pdf = pd.read_excel(
    "/Volumes/data_exploration_hub/marketing_hub/offline_data/CineClub Printed Gift Codes Dec 2025.xlsx",
    dtype=str,
)

df = spark.createDataFrame(pdf)
df.count()
```

```python
df.display()
```

```python
df_redemptions = spark.table("cpx_dataanalytics_silver.cineclub.gifting")

df_redemptions_status = (
    df
    .join(df_redemptions, F.col("Gift_Codes") == F.col("id"), "left")
    .select(
        "Gift_Codes",
        "is_redeemed",
        F.to_date("redemption_date").alias("redemption_date")
    )
)
```
