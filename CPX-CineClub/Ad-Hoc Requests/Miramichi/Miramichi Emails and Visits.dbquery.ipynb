```sql
SELECT Distinct (c.CDE_ID), plan_id, renewal_status from cpx_dataanalytics_gold.marketing.mkt_subscriber_profile_snapshot as c
LEFT Join cpx_dataanalytics_gold.customer360.c360_f_transaction as t
on c.cde_id = t.CDE_ID
Left Join cpx_dataanalytics_gold.customer360.cpx_d_location as L
on t.LocationId = L.LocationID
Where c.snapshot_date_id = '20260329'
AND t.VisitDateId > '20250101'
And L.Location_Name LIKE "%Miramichi%"
And c.renewal_status Like "Active%"
And c.plan_id Like "%annual%" OR plan_id like "%gift%"
GROUP BY c.CDE_ID, plan_id, renewal_status
```
