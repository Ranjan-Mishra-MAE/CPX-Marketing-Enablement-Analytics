```sql
SELECT
  VisitDateId DIV 10000                             AS visit_year,
  COUNT(DISTINCT CDE_ID)                            AS distinct_members,
  COUNT(DISTINCT CDE_ID, VisitDateId, LocationId)  AS total_visits,
  SUM(Quantity)                                     AS total_quantity
FROM cpx_dataanalytics_gold.customer360.c360_f_transaction
WHERE CineClubFLAG = 'Yes'
AND VisitDateId < 20260701
GROUP BY visit_year
ORDER BY visit_year
```
